```python
import pandas as pd
import numpy as np

employments=pd.read_csv('/Users/alexfion/Dropbox/Alex/dataViz/AShymanskaya.github.io/portfolio/visualizations/pacific_dataviz/full_story/data/employment.csv')
agricultural_land=pd.read_csv('/Users/alexfion/Dropbox/Alex/dataViz/AShymanskaya.github.io/portfolio/visualizations/pacific_dataviz/full_story/data/agricultural_land.csv')
food_security=pd.read_csv('/Users/alexfion/Dropbox/Alex/dataViz/AShymanskaya.github.io/portfolio/visualizations/pacific_dataviz/full_story/data/food_insecurity.csv')[['GEO_PICT','OBS_VALUE']].rename(columns={'OBS_VALUE':'food_insecurity'})
fishing=pd.read_csv('/Users/alexfion/Dropbox/Alex/dataViz/AShymanskaya.github.io/portfolio/visualizations/pacific_dataviz/full_story/data/fishing_total.csv')[['GEO_PICT','OBS_VALUE']].rename(columns={'OBS_VALUE':'fishing_families'})

```

```python
fishing.fishing_families.mean()
```

```python
employment_last_time = employments[employments['Indicator']=='Employment in agriculture (% of total employment) (modeled ILO estimate)'][['GEO_PICT','TIME_PERIOD','OBS_VALUE']].groupby('GEO_PICT').last().reset_index().drop(columns='TIME_PERIOD').rename(columns={'OBS_VALUE':'employment_agriculture'})

value_last_time = employments[employments['Indicator']=='Agriculture value added (% of GDP)'][['GEO_PICT','TIME_PERIOD','OBS_VALUE']].groupby('GEO_PICT').last().reset_index().drop(columns='TIME_PERIOD').rename(columns={'OBS_VALUE':'agriculture_value_added'})

agricultural_land_last_time = agricultural_land[['GEO_PICT','TIME_PERIOD','OBS_VALUE','Pacific Island Countries and territories']].groupby('GEO_PICT').last().reset_index().drop(columns='TIME_PERIOD').rename(columns={'OBS_VALUE':'agricultural_land'})

```

```python
result = agricultural_land_last_time.merge(value_last_time, on='GEO_PICT',how='left') \
           .merge(employment_last_time, on='GEO_PICT',how='left')\
           .merge(food_security, on='GEO_PICT',how='left')\
           .merge(fishing, on='GEO_PICT',how='left')\
           
#Agricultural Land Efficiency Vulnerability Index (ALEVI)
result['ALEVI']=result['agriculture_value_added']/result['agricultural_land']
#Agricultural Employment Vulnerability Index (AEVI)
result['AEVI']=result['employment_agriculture']/result['agricultural_land']

result.to_csv('/Users/alexfion/Dropbox/Alex/dataViz/AShymanskaya.github.io/portfolio/visualizations/pacific_dataviz/full_story/data/alevi.csv')
```
