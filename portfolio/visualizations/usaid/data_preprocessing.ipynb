```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Read the CSV file
df = pd.read_csv('us_foreign_aid_complete.csv')

# Print info about the data
print("Data shape:", df.shape)
print("\nColumns:", df.columns.tolist())

# Verify data in Country Name and US Category Name columns
print("\nUnique countries:", df['Country Name'].nunique())
print("Unique categories:", df['US Category Name'].nunique())

# Yearly contributions by sector
yearly_sector = df.groupby(['Fiscal Year', 'US Category Name'])['Constant Dollar Amount'].sum().reset_index()
print("\nYearly sector data shape:", yearly_sector.shape)

# Country contributions by sector and year
country_sector = df.groupby(['Country Name', 'US Category Name', 'Fiscal Year'])['Constant Dollar Amount'].sum().reset_index()
print("\nCountry sector data shape:", country_sector.shape)

# Print sample of country_sector data
print("\nSample of country_sector data:")
print(country_sector.head())

# Create sunburst with additional configurations
fig_country = px.sunburst(
    country_sector,
    path=['US Category Name', 'Country Name'],
    values='Constant Dollar Amount',
    title='Distribution by Sector and Country',
    maxdepth=2,  # Limit depth to improve visibility
    branchvalues='total'  # Use total values for better distribution
)

# Update layout for better visibility
fig_country.update_layout(
    width=1000,
    height=1000,
    uniformtext=dict(minsize=10, mode='hide')
)

# Create yearly stacked bar chart
fig_yearly = px.bar(
    yearly_sector, 
    x='Fiscal Year', 
    y='Constant Dollar Amount',
    color='US Category Name',
    title='Yearly Contributions by Sector',
    barmode='stack'
)

# Update layout for better readability
fig_yearly.update_layout(
    xaxis_title="Fiscal Year",
    yaxis_title="Constant Dollar Amount (USD)",
    legend_title="Sector",
    height=600
)

# Export aggregated data with formatting
yearly_sector_pivot = yearly_sector.pivot(
    index='Fiscal Year', 
    columns='US Category Name', 
    values='Constant Dollar Amount'
)
yearly_sector_pivot.to_csv('yearly_sector_summary.csv')
# Format country_sector data before saving
country_sector['Constant Dollar Amount'] = country_sector['Constant Dollar Amount'].round(2)
country_sector.to_csv('country_sector_summary.csv', index=False)

# Print the total aid amount by sector for verification
print("\nTotal aid by sector:")
print(country_sector.groupby('US Category Name')['Constant Dollar Amount'].sum().sort_values(ascending=False))
```
