```python
import pandas as pd
data=pd.read_csv('us_foreign_aid_usg_sector (1).csv')
```

```python
category_sectors = {
    cat: data[data['US Category Name'] == cat]['US Sector Name'].unique().tolist()
    for cat in data['US Category Name'].unique()
}
```

```python
category_sectors
```

```python
# Group by Category and Sector, then sum the amounts
grouped = data.groupby(['US Category Name', 'US Sector Name'])['current_amount'].sum().reset_index()

# Then, get the top 5 sectors by amount within each category
def get_top_5(group):
    return group.nlargest(5, 'current_amount')

# Group by Category and apply the top 5 function
top_sectors = grouped.groupby('US Category Name').apply(get_top_5).reset_index(drop=True)

# Sort the final result by Category and then by amount (descending)
top_sectors = top_sectors.sort_values(['US Category Name', 'current_amount'], ascending=[True, False])
```
